```python
	
###Connect to Context Broker

from ngsildclient import Client, Entity, iso8601

client = Client(hostname="100.85.126.64",port=1026)
```

```python
client.types.list()
```

```python

# Get all entities of type videoRecorded
client.query(type="videoRecorded")
```

```python
entities = client.query_head(type="videoRecorded")
for entity in entities:
    print(entity.id)
    print(entity.type)
    print(entity['inferred']['value'])
    print("-----")
```

```python
# Get last 10 entities of type videoRecorded
entities = client.query(type="videoRecorded")[-10:]
for entity in entities:
    if entity['inferred']['value'] == False:
        result = client.get(entity.id)
        print(result)
        
```

```python
#Get video path
entities = client.query(type="videoRecorded")
for entity in entities:
    if entity['inferred']['value'] == False:
        print(entity['path']['value'])
        
```

```python
#Change a given entity to true/false

entity = entities[0]

id = entity['id']
print(id)

entity = client.get(id)
entity['inferred']['value'] = False #can change to  true
print(entity)
```

```python
def build_entity(sensor: dict) -> Entity:
    e = Entity("TemperatureSensor", f"{sensor['id']}")
    e.obs()
    e.prop("temperature", sensor["temperature"])
    return e
```

```python
tempSensor = {"id": "001", "temperature": 17}
entity = build_entity(tempSensor)
print (entity)
#upser entity in client
client.upsert(entity)
```
