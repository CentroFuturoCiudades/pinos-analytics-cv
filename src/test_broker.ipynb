```python
import csv
import numpy as np
import cv2
import json

CAMARA_NUMBER_IN_RTSP = ADD_HERE  # Change this to the camera number you are working with
CAMARA_NUMBER_IN_HIK_VISION = ADD_HERE  # Change this to the camera number in Hik Vision (aka the camara number shown in the video)

# Add camera csv file as input
input_file = f'../csv/points_cam0{CAMARA_NUMBER_IN_HIK_VISION}.csv'

source_list = []  # For Camera coordinates
target_list = []  # For Birdseye coordinates

# Read CSV file
with open(input_file, mode='r') as infile:
    reader = csv.reader(infile)
    
    # Skip the header row
    next(reader)
    
    for row in reader:
        birdseye_coords = row[1].strip('"')
        camera_coords = row[2].strip('"')

        # Check if both coordinates are non-empty
        if not birdseye_coords or not camera_coords:
            print(f"Skipping row with missing coordinates: {row}")
            continue

        # Convert the string "(x, y)" into a tuple (x, y)
        birdseye_tuple = tuple(map(float, birdseye_coords.strip('()').split(',')))
        camera_tuple = tuple(map(float, camera_coords.strip('()').split(',')))
            
        target_list.append(birdseye_tuple)
        source_list.append(camera_tuple)

print("Source List (Camera):")
print(source_list)

print("\nTarget List (Birdseye):")
print(target_list)

"""Calculate the homography from source to target using RANSAC."""
source_list = np.array(source_list).astype(np.float32).reshape(-1, 1, 2)
target_list = np.array(target_list).astype(np.float32).reshape(-1, 1, 2)
h, status = cv2.findHomography(source_list, target_list)

with engine.begin() as conn:
    # Insert into detectionsobserved table if ID does not already exist
    conn.execute(sa.text("""
        INSERT INTO camara_homography_matrices (camera_number, matrix)
        VALUES (:camera_number, :matrix)
        ON CONFLICT (camera_number) DO NOTHING
    """), {
        "camera_number": CAMARA_NUMBER_IN_RTSP,
        "matrix": json.dumps(h.tolist()),
     })

print("Homography Matrix:")
print(json.dumps(h.tolist()))
```

```python
import numpy as np
from scipy.interpolate import interp1d
import datetime
import json
import pandas

# Get all detections of id 1 from video camera5_2025_05_28-12_40_57_PM.mp4
detections = pd.read_sql("SELECT * FROM detectionsobserved WHERE video_path = '2025_05_23-03_54_50_PM.mp4' AND detection_id = '29'", engine)
#print(detections['skeleton'][16])

times = detections['timestamp'].values
times_secs = (times - times[0]) / np.timedelta64(1, 's')

#x_coordinates = [coord[0] for coord in coordinates]
#y_coordinates = [coord[1] for coord in coordinates]

# Extract skeletons and get x/y for keypoint 16 (example)
x_coordinates = []
y_coordinates = []

i = 0
for skeleton in detections['skeleton']:
    x_coordinates.append(skeleton[16][0])  # x coordinate of keypoint 16
    y_coordinates.append(skeleton[16][1])  # y coordinate of keypoint 16

print("Times:", times_secs)
print("X Coordinates:", x_coordinates)
print("Y Coordinates:", y_coordinates)

print (len(times_secs), len(x_coordinates), len(y_coordinates))

# Data frame to remove (0,0) points - will not need in new version?
# Create a DataFrame
df = pd.DataFrame({
    'timestamp': times,
    'time_secs': times_secs,
    'x': x_coordinates,
    'y': y_coordinates
})

# Filter out (0,0) points first
df = df[(df['x'] != 0) | (df['y'] != 0)]
#

# Keep only the first occurrence for each timestamp
df = df.drop_duplicates(subset=['time_secs'], keep='first')

# Sort by time (important for interpolation)
df = df.sort_values('time_secs')

# Extract cleaned data
times_secs_clean = df['time_secs'].values
x_coordinates_clean = df['x'].values
y_coordinates_clean = df['y'].values

print("Cleaned Times:", times_secs_clean)
print("Cleaned X Coordinates:", x_coordinates_clean)
print("Cleaned Y Coordinates:", y_coordinates_clean)

# Create interpolation functions
f_x = interp1d(times_secs_clean, x_coordinates_clean, kind='linear')  # Linear interpolation for x
f_y = interp1d(times_secs_clean, y_coordinates_clean, kind='linear') # Linear interpolation for y

# Check if times_secs has more than one unique value
if np.ptp(times_secs) == 0:
    print("All timestamps are the same. Interpolation is not possible.")
else:
    # Interpolate only within the valid range
    min_time = np.min(times_secs)
    max_time = np.max(times_secs)
    interp_point = min_time + 1.5 # example

    interp_x = f_x(interp_point)
    interp_y = f_y(interp_point)

    print(f"Interpolated X at t={interp_point}: {interp_x}")
    print(f"Interpolated Y at t={interp_point}: {interp_y}")


```
